## Conversion of moles to mass fraction

Now let's see how we can use Python to solve a simple CP101 problem: calculating the mass fraction of a stream, given the number of moles that are in it.

Consider a stream containing $1\,{\rm mol}$ of methane with a stochiometric amount of air to fully combust it.  The stream will consist of methane (${\rm CH}_4$),
oxygen (${\rm O}_2$), and nitrogen (${\rm N}_2$).

Let's first determine the number of moles of each component in the system.  We know there is $1\,{\rm mol}$ of methane.  The combustion reacttion of methane can be written as
$$
{\rm CH}_4 + 2 {\rm O}_2 \rightleftharpoons {\rm CO}_2 + 2{\rm H_2O}
$$

So we see that for every mole of methane, we need $2\,{\rm mol}$ of oxygen.  Air consists of $21\,{\rm mol}\%$ of oxygen and $79\,{\rm mol}\%$ of nitrogen.

```python
print('moles of methane:',1)
print('moles of oxygen:', 1*2)
print('moles of nitrogen:', 1*2*79/21)
```

The total number of moles in the stream is:

```python
1 + 2 + 2*79/21
```

Now we can determine the mass of each component in the stream, which will be number of moles of the component multiplied by its molecular weight.

For methane, this is:

```python
1 * 16
```

For oxygen, this is:

```python
(1*2) * 32
```

For nitrogen, this is:

```python
(2*79/21) * 28
```

To determine the mass fractions, we need to know the total mass of the stream.  This is just the sum of the mass of each of the components in the stream:

```python
1*16 + (2)*32 + (2*79/21)*28
```

Once we have the total mass, we can then report the mass fraction of each component by just dividing its mass by the total mass.

```python
print('mass fraction of methanne:', 1*16/290.66666666666663)
print('mass fraction of oxygen:', 2*32/290.66666666666663)
print('mass fraction of nitrogen:', (2*79/21)*28/290.66666666666663)
```

### Use of lists

In the calculations so far, Python was simply used as a replacement for a calculator.  While this is fine, in terms of how you would normally do calculations on paper, it limits the reusability and extensibility of the code.  

In the following, we look to use variables to generalize the calculations.

```python
N_CH4 = 1               # moles of methane in stream
N_O2 = N_CH4 * 2        # moles of oxygen
N_N2 = N_O2 * 79/21     # moles of nitrogen

Mw_CH4 = 16             # molecular weight of methane
Mw_O2 = 31              # molecular weight of oxygen
Mw_N2 = 28              # molecular weight of nitrogen

m_CH4 = N_CH4 * Mw_CH4  # mass of methane
m_O2 = N_O2 * Mw_O2     # mass of oxygen
m_N2 = N_N2 * Mw_N2     # mass of nitrogen
print('mass of methane', N_CH4)
print('mass of oxygen', N_O2)
print('mass of nitrogen', N_N2)
m_total = m_CH4 + m_O2 + m_N2   # total mass of stream
print('total mass:', m_total)

w_CH4 = m_CH4 / m_total
w_O2 = m_O2 / m_total
w_N2 = m_N2 / m_total
print('weight fraction of methane', w_CH4)
print('weight fraction of oxygen', w_O2)
print('weight fraction of nitrogen', w_N2)

```

By using variables, we can easily redo calculations by just changing the values of the different variables.  For example, we can choose to use more precise values for the molecular weights of [methane](https://webbook.nist.gov/cgi/cbook.cgi?ID=C74828&Mask=1#Thermo-Gas), [oxygen](https://webbook.nist.gov/cgi/cbook.cgi?ID=C7782447&Mask=1#Thermo-Gas), and [nitrogen](https://webbook.nist.gov/cgi/cbook.cgi?ID=C7727379&Units=SI&Mask=1#Thermo-Gas) from the [NIST chemistry webbook](https://webbook.nist.gov/chemistry/).

```python
N_CH4 = 1               # moles of methane in stream
N_O2 = N_CH4 * 2        # moles of oxygen
N_N2 = N_O2 * 79/21     # moles of nitrogen

Mw_CH4 = 16.0425        # molecular weight of methane
Mw_O2 = 31.9988         # molecular weight of oxygen
Mw_N2 = 28.0134         # molecular weight of nitrogen

m_CH4 = N_CH4 * Mw_CH4  # mass of methane
m_O2 = N_O2 * Mw_O2     # mass of oxygen
m_N2 = N_N2 * Mw_N2     # mass of nitrogen
print('mass of methane', N_CH4)
print('mass of oxygen', N_O2)
print('mass of nitrogen', N_N2)
m_total = m_CH4 + m_O2 + m_N2   # total mass of stream
print('total mass:', m_total)

w_CH4 = m_CH4 / m_total
w_O2 = m_O2 / m_total
w_N2 = m_N2 / m_total
print('weight fraction of methane', w_CH4)
print('weight fraction of oxygen', w_O2)
print('weight fraction of nitrogen', w_N2)
```

### Use of lists

The use of variables allows some reusability within the code when we only want to change the value of a variable.  It becomes more problematic when we want to make changes such as altering the number of components in the system.  Lists allow us to "easily" alter the number of components in the system.


```python
# component 0: methane
# component 1: oxygen
# component 2: nitrogen

N_list = [1, 2, 2*79/21]                 # list of mole numbers
Mw_list = [16.0425, 31.9988, 28.0134]    # list of molecular weights

print('number of species:', len(N_list))

mass_total = 0
for i in range(len(N_list)):
    # i is the index of the list, and indicates the particular component
    m = N_list[i] * Mw_list[i]   # This is the mass of species i, which is the product of the number of
                                 # moles of species i multiplied by the molecular weight of species i
    mass_total = mass_total + m  # increment the total of mass of the stream
#    print(i, m, mass_total)
print('total mass:', mass_total)

print()
print('mass fractions')
for i in range(len(N_list)):
    m = N_list[i] * Mw_list[i]
    print('species:', i, m/mass_total)

```

To increase the number of species in the system, we simply add their mole numbers to `N_list` and their molecular weights to `Mw_list`.

```python
# component 0: methane
# component 1: oxygen
# component 2: nitrogen
# component 3: carbon dioxide
# component 4: water

N_list = [1, 2, 2*79/21, 0.5, 0.1]              # mole nunmbers
Mw_list = [16.0425, 31.9988, 28.0134, 44, 18]   # molecular weights

print('number of species:', len(N_list))

mass_total = 0
for i in range(len(N_list)):
    # i is the index of the list, and indicates the particular component
    m = N_list[i] * Mw_list[i]   # This is the mass of species i, which is the product of the number of
                                 # moles of species i multiplied by the molecular weight of species i
    mass_total = mass_total + m  # increment the total of mass of the stream
#    print(i, m, mass_total)
print('total mass:', mass_total)

print()
print('mass fractions')
for i in range(len(N_list)):
    m = N_list[i] * Mw_list[i]
    print('species:', i, m/mass_total)
```

## Use of dictionaries

The use of lists is a good way to make the number of components in the system easily generalizable; however, it comes with the inconvenience that the actual identity of the components in the list is not explicitly clear.  The order in which values appear in the list is important and must be carefully entered.

Dictionaries eliminates these problems of lists, while maintaining their flexibility.  With dictionaries, we are free to create more descriptive keys to identify which components we are referring to.  Below, we reimplement the code using dictionaries.

```python
N_dict = {'CH4': 1, 'O2': 2, 'N2': 2*79/21, 'H2O': 0.5}
Mw_dict = {'CH4': 16.0425, 'O2': 31.9988, 'N2': 28.0134, 'H2O': 18}

total_mass = 0
for molecule, moles in N_dict.items():
    mass = moles * Mw_dict[molecule]
    total_mass = total_mass + mass

print('mass fractions')
w_dict = {}
for molecule, moles in N_dict.items():
    mass = moles * Mw_dict[molecule]
    w_dict[molecule] = mass/total_mass
    print(molecule, mass/total_mass)

print(w_dict)
```
