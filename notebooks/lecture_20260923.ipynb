## Example from CP101

Hydrogen is generally produced from water via two different methods. The first is one is
through the water gas shift reaction:
$$
{\rm H_2O(g) + CO(g) \leftrightharpoons H_2(g) + CO_2(g)}
$$

The second method, is the electrolysis of water:
$$
{\rm H_2O(g) = H_2(g) + \frac{1}{2}O_2(g)}
$$

However, these reactions are generally carried out at different
temperatures. Calculate the reaction enthalpies for the water gas shift and electrolysis reactions at $600^\circ{\rm C}$ and $900^\circ{\rm C}$, respectively.  Is there any temperature at which
these two reactions require the same amount of enthalpy to produce
hydrogen?


The molar heat capacity $C_p$ can be estimated using the equation
$$
C_p = a + b T
$$
where $T$ is the absolute temperature, and $a$ and $b$ are parameters that are listed for various substances in the table below.


| Species | $\Delta H_f$ | $a$ |  $b$ |
| :------- | ----: | ----: | ----: |
|         | ${\rm kJ\,mol^{-1}}$ | ${\rm J\,mol^{-1}\,K^{-1}}$ | ${\rm J\,mol^{-1}\,K^{-2}}$ |
| ${\rm H_2O(g)}$ | $ -242.20$ | $32.47$|  $0.0086$ |
| ${\rm H_2(g) }$ | $0.00$ | $ 27.32$ | $0.0033$ |
| ${\rm CO(g) }$ | $ -110.70$ | $ 29.61$ | $0.0030$ |
| ${\rm CO_2(g)}$ | $-394.09$ | $ 44.32$| $0.0073 $|
| ${\rm O_2(g) }$ | $ 0.00$ | $ 30.51$ | $0.0035 $|


### Inputting data

If we want to use Python to help do the required calculations, we first need to enter the data into a variable so that the necessary information can be easily retrieved.  In this case, we will use a dictionary called `data`.  Each entry in the dictionary `data` is the information for a particular compound.  It is a "dictionary of dictionaries".

```python
data = {}
data['H2O'] = {'Hf': -242.20, 'a': 32.47, 'b': 0.0086}     # information for H2O
data['H2']  =	{'Hf':    0.00, 'a': 27.32, 'b': 0.0033}     # information for H2
data['CO']  = {'Hf': -110.70, 'a': 29.61, 'b': 0.0030}     # information for CO
data['CO2'] = {'Hf': -394.09, 'a': 44.32, 'b': 0.0073}     # information for CO2
data['O2']  =	{'Hf':    0.00, 'a': 30.51, 'b': 0.0035}     # information for O2


print(data)
print(data['H2O'])
print(data['H2O']['Hf'])

mol = 'CO2'
print(data[mol])
print(data[mol]['Hf'])

for mol in ['H2O', 'CO']:
  print(mol, data[mol]['Hf'])

```

In order to compute the enthalpy of the water gas shift reaction, we now construct a dictionary called `reaction` that contains the stoichiometric coefficients of the reaction.

If we loop over this dictionary and sum the enthalpy of formation contribution for each species, we can then get the enthalpy of reaction at standard conditions (i.e. $25^\circ{\rm C}$ and $1\,{\rm bar}$).

```python
reaction = {'H2O':-1, 'CO':-1, 'H2':1, 'CO2':1}
print(reaction)
print(reaction['H2O'])


Hrxn = 0  # zero the accumulator, which will eventually be the enthalpy of reaction
for mol, nu in reaction.items():
 # print(mol, stoichiometric_coefficient)
  H = data[mol]['Hf']
  Hrxn = Hrxn + nu * H
  print(f'{mol}:  {H} kJ/mol')
print(f'enthalpy of reaction = {Hrxn:.2f} kJ/mol')
#
```

Note that we can easily reuse this code to calculate the enthalpy of the water splitting reaction by just changing the `reaction` dictionary.

```python

reaction = {'H2O':-1, 'H2':1, 'O2':0.5}


Hrxn = 0  # zero the accumulator, which will eventually be the enthalpy of reaction
for mol, nu in reaction.items():
 # print(mol, stoichiometric_coefficient)
  H = data[mol]['Hf']
  Hrxn = Hrxn + nu * H
  print(f'{mol}:  {H} kJ/mol')
print(f'enthalpy of reaction = {Hrxn:.2f} kJ/mol')
```

We have now calculated the enthalpy of reaction at standard conditions, however, we want to calculate it at different temperatures.  To do that, we need to know how the enthalpy of each component changes with temperature.  We can do this through the heat capacity.

The enthalpy change with temperature is directly related to the heat capacity as
$$
\begin{align*}
dH &= C_p dT
\\
H(T) - H(T_0)
&=
\int_{T_0}^T dT\,C_p(T)
\\
&=
\int_{T_0}^T dT\,(a+bT)
\\
&= \left.a T + \frac{b}{2} T^2\right]_{T_0}^T
\\
&= a(T-T_0) + \frac{b}{2} (T^2-T_0^2)
\end{align*}
$$

If we take $T_0=25^\circ{\rm C}$, then we find:
$$
H(T) = \Delta H_f
+ a(T-T_0) + \frac{b}{2} (T^2-T_0^2)
$$

Below, we modify our code to take into account the temperature dependence of the enthalpy.

```python

#print(reaction)
#print(reaction['H2O'])

T0 = 25 + 273.15

print('water gas shift reaction')
reaction = {'H2O':-1, 'CO':-1, 'H2':1, 'CO2':1}
T = 600 + 273.15
Hrxn = 0
for mol, nu in reaction.items():
 # print(mol, stoichiometric_coefficient)
  Hf = data[mol]['Hf']*1000
  a = data[mol]['a']
  b = data[mol]['b']
  H = Hf + a*(T-T0) + 0.5*b*(T**2-T0**2)    # Be careful with units!!!
  Hrxn = Hrxn + nu * H
  print(mol, H)

print(f'enthalpy of reaction = {Hrxn/1000:.2f} kJ/mol')

print()
print('water splitting')
reaction = {'H2O':-1, 'H2':1, 'O2':0.5}
T = 900 + 273.15
Hrxn = 0
for mol, nu in reaction.items():
 # print(mol, stoichiometric_coefficient)
  Hf = data[mol]['Hf']*1000
  a = data[mol]['a']
  b = data[mol]['b']
  H = Hf + a*(T-T0) + 0.5*b*(T**2-T0**2)    # Be careful with units!!!
  Hrxn = Hrxn + nu * H
  print(mol, H)

print(f'enthalpy of reaction = {Hrxn/1000:.2f} kJ/mol')
```

Now lets run this code at different temperatures
